# Task 1: Manually calculate evaluation metrics

```python
# Given values
TP = 50
TN = 120
FP = 10
FN = 20

# Calculate Accuracy
accuracy = (TP + TN) / (TP + TN + FP + FN)

# Calculate Precision
precision = TP / (TP + FP)

# Calculate Recall
recall = TP / (TP + FN)

# Calculate F1 Score
f1_score = 2 * (precision * recall) / (precision + recall)

# Print results
print("Accuracy:", accuracy)
print("Precision:", precision)
print("Recall:", recall)
print("F1 Score:", f1_score)

# Print results in percentage
print("\nAccuracy:", accuracy * 100, "%")
print("Precision:", precision * 100, "%")
print("Recall:", recall * 100, "%")
print("F1 Score:", f1_score * 100, "%")
```

# Task 2: Create a confusion matrix function

```python
def print_confusion_matrix(tp, tn, fp, fn):

    print("\nConfusion Matrix")
    print("-------------------------------")
    print("                 Predicted")
    print("               Positive  Negative")
    print("--------------------------------")
    print("Actual Positive    ", tp, "      ", fn)
    print("Actual Negative    ", fp, "      ", tn)
    print("--------------------------------")

# Call function
print_confusion_matrix(50, 120, 10, 20)
```
