### Task 1 – StandardScaler on IPL Dataset

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import MinMaxScaler
```

```python
data = {
    "player": ["Virat", "Rohit", "Bumrah", "Hardik", "Rahul",
               "Jadeja", "Gill", "Pant", "Surya", "Siraj"],
    "runs": [639, 417, 15, 346, 520, 189, 890, 446, 605, 10],
    "wickets": [0, 0, 20, 11, 0, 10, 0, 0, 0, 14],
    "matches": [15, 14, 13, 14, 14, 16, 17, 13, 16, 12],
    "strike_rate": [139, 150, 85, 151, 136, 142, 157, 155, 167, 90]
}

ipl_df = pd.DataFrame(data)

display(ipl_df)
```

```python
# Select numeric columns
numeric_columns = ["runs", "wickets", "matches", "strike_rate"]

# Create StandardScaler
scaler = StandardScaler()

# Scale numeric features
scaled_data = scaler.fit_transform(ipl_df[numeric_columns])

# Create new DataFrame
scaled_df = pd.DataFrame(scaled_data, columns=numeric_columns)

print("Scaled IPL Dataset:")
display(scaled_df)
```

```python
# Save scaled data into CSV file
scaled_df.to_csv("IPL_Scaled_Data.csv", index=False)

print("Scaled dataset saved successfully!")
```

### Task 2 – MinMaxScaler on Flipkart Product Ratings

```python
data = {
    "price": [500, 1200, 2500, 800, 3000, 1500, 4000, 600],
    "rating": [4.2, 3.8, 4.5, 4.0, 4.7, 3.5, 4.8, 4.1],
    "reviews": [100, 250, 500, 150, 800, 200, 1000, 120],
    "discount": [10, 20, 15, 25, 30, 10, 40, 5]
}

flipkart_df = pd.DataFrame(data)

print("Original Dataset:")
display(flipkart_df)
```

```python
print("Minimum Values Before Scaling:")
print(flipkart_df.min())

print("\nMaximum Values Before Scaling:")
print(flipkart_df.max())
```

```python
# Create MinMaxScaler
scaler = MinMaxScaler()

# Apply scaling
scaled_data = scaler.fit_transform(flipkart_df)

# Create scaled DataFrame
flipkart_scaled = pd.DataFrame(
    scaled_data,
    columns=flipkart_df.columns
)

print("Scaled Dataset:")
display(flipkart_scaled)
```

```python
print("Minimum Values After Scaling:")
print(flipkart_scaled.min())

print("\nMaximum Values After Scaling:")
print(flipkart_scaled.max())
```

### Task 3 – Correlation Analysis on Spotify Dataset

```python
spotify_data = {
    "danceability": [0.80, 0.60, 0.90, 0.45, 0.75,
                     0.55, 0.85, 0.70, 0.65, 0.95],
    "energy": [0.85, 0.70, 0.95, 0.40, 0.80,
               0.60, 0.90, 0.75, 0.50, 0.98],
    "tempo": [120, 100, 130, 90, 115, 105, 140, 125, 95, 150],
    "popularity": [90, 65, 95, 40, 85, 55, 92, 78, 45, 98],
    "acousticness": [0.10, 0.30, 0.05, 0.80, 0.15,
                     0.60, 0.08, 0.20, 0.75, 0.02]
}

spotify_df = pd.DataFrame(spotify_data)

display(spotify_df)
```

```python
# Calculate correlation
correlation = spotify_df.corr()

print("Correlation with Popularity:")
print(correlation["popularity"].sort_values(ascending=False))
```

```python
# Select top 2 features related to popularity
top_features = correlation["popularity"].drop("popularity").abs().nlargest(2)

print("Top 2 Features:")
print(top_features)
```

```python
top_features.plot(kind="bar", figsize=(7, 4))

plt.title("Top 2 Features Related to Popularity")
plt.xlabel("Features")
plt.ylabel("Absolute Correlation")
plt.show()
```

### Task 4 – Recursive Feature Elimination (RFE)

```python
zomato_data = {
    "rating": [4.5, 3.8, 4.2, 4.8, 3.5, 4.0, 4.6, 3.9, 4.3, 4.7],
    "votes": [500, 200, 350, 800, 100, 250, 650, 180, 450, 700],
    "distance": [2, 5, 3, 1, 7, 4, 2, 6, 3, 1],
    "delivery_time": [25, 40, 30, 20, 50, 35, 25, 45, 30, 22],
    "menu_items": [50, 30, 45, 80, 20, 35, 65, 25, 55, 75],
    "average_cost": [800, 400, 600, 1200, 300, 500, 1000, 350, 750, 1100]
}

zomato_df = pd.DataFrame(zomato_data)

display(zomato_df)
```

```python
from sklearn.feature_selection import RFE
from sklearn.tree import DecisionTreeRegressor

# Select input features
X = zomato_df.drop("average_cost", axis=1)

# Select target variable
y = zomato_df["average_cost"]

# Create Decision Tree model
model = DecisionTreeRegressor(random_state=42)

# Apply RFE to select 3 features
rfe = RFE(estimator=model, n_features_to_select=3)

rfe.fit(X, y)

# Display selected features
selected_features = X.columns[rfe.support_]

print("Selected 3 Features:")
print(selected_features.tolist())
```

```python
ranking = pd.DataFrame({
    "Feature": X.columns,
    "Ranking": rfe.ranking_
})

print("Feature Ranking:")
display(ranking.sort_values("Ranking"))
```

### Task 5 – Detect and Remove Outliers Using IQR

```python
bookmyshow_data = {
    "movie_name": [
        "Movie A", "Movie B", "Movie C", "Movie D", "Movie E",
        "Movie F", "Movie G", "Movie H", "Movie I", "Movie J",
        "Movie K", "Movie L"
    ],
    "user_ratings": [4.2, 3.8, 4.5, 4.0, 3.9, 4.3,
                     4.1, 4.4, 3.7, 4.2, 0.5, 9.8]
}

bookmyshow_df = pd.DataFrame(bookmyshow_data)

print("Original Dataset:")
display(bookmyshow_df)
```

```python
# Calculate Q1 and Q3
Q1 = bookmyshow_df["user_ratings"].quantile(0.25)
Q3 = bookmyshow_df["user_ratings"].quantile(0.75)

# Calculate IQR
IQR = Q3 - Q1

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
```

```python
# Calculate limits
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
```

```python
# Remove outliers
cleaned_df = bookmyshow_df[
    (bookmyshow_df["user_ratings"] >= lower_limit) &
    (bookmyshow_df["user_ratings"] <= upper_limit)
]

# Calculate dropped rows
dropped_rows = len(bookmyshow_df) - len(cleaned_df)

print("Original Rows:", len(bookmyshow_df))
print("Rows After Removing Outliers:", len(cleaned_df))
print("Dropped Rows:", dropped_rows)

print("\nCleaned Dataset:")
display(cleaned_df)
```
