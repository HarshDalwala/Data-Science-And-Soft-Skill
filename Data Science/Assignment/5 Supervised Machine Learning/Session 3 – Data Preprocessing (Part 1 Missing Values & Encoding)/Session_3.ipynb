```python
df = pd.read_csv("ipl_sample_missing_values.csv")

print("First 5 Rows:")
display(df.head())
```

```python
print("Missing Values in Each Column:")
print(df.isnull().sum())
```

### Task 2 – Fill Missing Player Age with Median

```python
# Calculate median age
median_age = df["player_age"].median()

# Fill missing ages with median
df["player_age"] = df["player_age"].fillna(median_age)

# Display updated column
print("Updated Player Age:")
print(df["player_age"])
```

### Task 3 – Fill Missing Team Names with Unknown

```python
# Replace missing team names
df["team"] = df["team"].fillna("Unknown")

# Display first 10 rows
print("Updated Team Column:")
display(df.head(10))
```

### Task 4 – One-Hot Encoding on Venue Column

```python
# Fill missing venue values
df["venue"] = df["venue"].fillna("Unknown")

# Apply One-Hot Encoding
df_encoded = pd.get_dummies(df, columns=["venue"], dtype=int)

# Display updated DataFrame
display(df_encoded.head(10))
```

### Task 5 – Label Encoding on Player Role

```python
from sklearn.preprocessing import LabelEncoder

# Fill missing player roles
df["player_role"] = df["player_role"].fillna("Unknown")

# Create LabelEncoder
le = LabelEncoder()

# Encode player roles
df["player_role_encoded"] = le.fit_transform(df["player_role"])

# Display mapping
print("Player Role Mapping:")

for role, value in zip(le.classes_, le.transform(le.classes_)):
    print(role, "=", value)

# Display updated DataFrame
display(df[["player_name", "player_role", "player_role_encoded"]])
```
