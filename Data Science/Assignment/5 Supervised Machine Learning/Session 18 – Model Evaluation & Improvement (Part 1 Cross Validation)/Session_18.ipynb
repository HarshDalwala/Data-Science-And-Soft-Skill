# Task 1 – 5-Fold Cross-Validation using KNN

```python
from sklearn.datasets import load_iris
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import cross_val_score

# Load Iris dataset
iris = load_iris()

X = iris.data
y = iris.target

# Create KNN model
knn = KNeighborsClassifier(n_neighbors=5)

# Perform 5-fold cross-validation
scores = cross_val_score(knn, X, y, cv=5)

print("Accuracy of Each Fold:", scores)
print("Average Accuracy:", scores.mean())
```

# Task 2 – 10-Fold Cross-Validation using Decision Tree

```python
from sklearn.datasets import load_breast_cancer
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import cross_val_score

# Load Breast Cancer dataset
data = load_breast_cancer()

X = data.data
y = data.target

# Create Decision Tree model
dt_model = DecisionTreeClassifier(random_state=42)

# Perform 10-fold cross-validation
scores = cross_val_score(dt_model, X, y, cv=10)

# Print accuracy of each fold
for i, score in enumerate(scores):
    print("Fold", i + 1, "Accuracy:", score)

print("\nAverage Accuracy:", scores.mean())
```

# Task 3 – StratifiedKFold on Iris Dataset

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import StratifiedKFold
import pandas as pd

# Load Iris dataset
iris = load_iris()

X = iris.data
y = iris.target

# Create StratifiedKFold
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

# Display class distribution in each fold
for fold, (train_index, test_index) in enumerate(skf.split(X, y)):

    print("\nFold", fold + 1)

    train_counts = pd.Series(y[train_index]).value_counts().sort_index()
    test_counts = pd.Series(y[test_index]).value_counts().sort_index()

    print("Training Class Counts:")
    print(train_counts)

    print("Testing Class Counts:")
    print(test_counts)
```

# Task 4 – Compare CV Values using Random Forest

```python
from sklearn.datasets import load_wine
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import cross_val_score
import pandas as pd

# Load Wine dataset
wine = load_wine()

X = wine.data
y = wine.target

# Create Random Forest model
rf_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

# Different CV values
cv_values = [3, 5, 10]

results = []

for cv in cv_values:

    scores = cross_val_score(rf_model, X, y, cv=cv)

    results.append({
        "CV Value": cv,
        "Mean Accuracy": scores.mean(),
        "Standard Deviation": scores.std()
    })

    print("\nCV =", cv)
    print("Fold Accuracies:", scores)
    print("Mean Accuracy:", scores.mean())
    print("Standard Deviation:", scores.std())

# Display comparison
results_df = pd.DataFrame(results)

print("\nFinal Comparison:")
print(results_df)

# Find most stable CV value
best_cv = results_df.loc[
    results_df["Standard Deviation"].idxmin()
]

print("\nMost Stable CV Value:", best_cv["CV Value"])
```

# Task 5 – Create a Reusable Cross-Validation Function with AI Help

```python
from sklearn.model_selection import cross_val_score

def evaluate_model(model, X, y, cv=5):

    scores = cross_val_score(model, X, y, cv=cv)

    mean_score = scores.mean()
    std_score = scores.std()

    return mean_score, std_score


# Load Iris dataset
from sklearn.datasets import load_iris
from sklearn.ensemble import RandomForestClassifier

iris = load_iris()

X = iris.data
y = iris.target

# Create classifier
model = RandomForestClassifier(n_estimators=100, random_state=42)

# Evaluate model
mean_accuracy, std_accuracy = evaluate_model(model, X, y, cv=5)

print("Mean Accuracy:", mean_accuracy)
print("Standard Deviation:", std_accuracy)
```
