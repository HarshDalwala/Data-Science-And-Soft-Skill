Instagram uses Supervised Learning to detect spam comments and unwanted messages. It learns from previously labelled spam and normal messages.

2. Spotify:

3. Mobile Games:

## Task 2: Classification of Machine Learning

1. Netflix recommending movies based on watch history.

Answer: Unsupervised Learning – It identifies patterns in users' watching habits and recommends similar movies.

2. Spotify grouping similar songs into playlists.

Answer: Unsupervised Learning – It groups songs based on similar features.

3. A self-driving car learning to park by trial and error.

Answer: Reinforcement Learning – It learns through actions, rewards, and mistakes.

```python
import matplotlib.pyplot as plt

steps = ["Data", "Preprocessing", "Modeling", "Evaluation", "Deployment"]

descriptions = [
    "Collect customer orders and ratings",
    "Clean data and remove errors",
    "Train model to predict delivery time",
    "Check prediction accuracy",
    "Use model in food delivery app"
]

plt.figure(figsize=(16, 5))

for i in range(5):
    plt.text(i * 3, 1, steps[i],
             ha='center',
             fontsize=11,
             bbox=dict(boxstyle='round,pad=0.8',
                       facecolor='lightblue',
                       edgecolor='black'))

    plt.text(i * 3, 0.4, descriptions[i],
             ha='center',
             fontsize=9)

    if i < 4:
        plt.annotate(
            '',
            xy=(i * 3 + 2, 1),
            xytext=(i * 3 + 1, 1),
            arrowprops=dict(arrowstyle='->', lw=2)
        )

plt.title("Machine Learning Workflow - Food Delivery App")
plt.axis('off')
plt.show()
```

## Task 4: Traditional Programming vs Machine Learning

Traditional Programming:

In traditional programming, we write fixed rules to identify spam messages. For example, if a message contains words like "Win Money" or "Free Prize", the program marks it as spam. It follows the rules written by the programmer.

Machine Learning:

In machine learning, we provide the computer with many examples of spam and normal messages. The model learns patterns from the data and predicts whether a new message is spam or not spam.

Conclusion:

Traditional programming depends on fixed rules, whereas machine learning learns from data and makes predictions.
