### Task 1 – Create Instagram Usage and Battery Data

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Instagram usage hours per day
instagram_hours = [1, 2, 3, 4, 5, 6, 7]

# Battery percentage at the end of each day
battery_percentage = [90, 80, 75, 65, 55, 45, 35]

print("Instagram Hours:", instagram_hours)
print("Battery Percentage:", battery_percentage)
```

```python
plt.scatter(instagram_hours, battery_percentage)

plt.xlabel("Instagram Usage Hours")
plt.ylabel("Battery Percentage")
plt.title("Instagram Usage vs Battery Percentage")

plt.show()
```

### Task 2 – Train a Linear Regression Model

```python
from sklearn.linear_model import LinearRegression

# Convert data into NumPy arrays
X = np.array(instagram_hours).reshape(-1, 1)
y = np.array(battery_percentage)

# Create Linear Regression model
model = LinearRegression()

# Train the model
model.fit(X, y)

# Print coefficient and intercept
print("Slope (Coefficient):", model.coef_[0])
print("Intercept:", model.intercept_)
```

### Task 3 – Visualize Regression Line

```python
# Predict battery percentage
predicted_battery = model.predict(X)

# Plot actual data
plt.scatter(X, y, label="Actual Data")

# Plot regression line
plt.plot(X, predicted_battery, color="red",
         label="Regression Line")

plt.xlabel("Instagram Usage Hours")
plt.ylabel("Battery Percentage")
plt.title("Linear Regression")

plt.legend()
plt.grid(True)

plt.show()
```

### Task 4 – Calculate Mean Squared Error (MSE)

```python
from sklearn.metrics import mean_squared_error

# Calculate MSE
mse = mean_squared_error(y, predicted_battery)

print("Mean Squared Error (MSE):", mse)
```
