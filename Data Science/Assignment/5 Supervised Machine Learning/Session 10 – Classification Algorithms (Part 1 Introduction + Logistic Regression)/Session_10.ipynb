### Task 1: Create a Sigmoid Function

```python
import numpy as np

# Create sigmoid function
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# Test sigmoid function
print("Sigmoid(-2):", sigmoid(-2))
print("Sigmoid(0):", sigmoid(0))
print("Sigmoid(3):", sigmoid(3))
```

### Task 2: Logistic Regression for Instagram Viral Posts

```python
import pandas as pd

from sklearn.linear_model import LogisticRegression

# Create Instagram dataset
data = {
    "likes": [100, 200, 500, 800, 1200,
              1500, 2000, 2500, 3000, 4000],

    "has_caption": [0, 1, 0, 1, 1,
                    0, 1, 1, 0, 1],

    "viral": [0, 0, 0, 0, 1,
              0, 1, 1, 1, 1]
}

df = pd.DataFrame(data)

print("Instagram Dataset:")
print(df)

# Features and target
X = df[["likes", "has_caption"]]
y = df["viral"]

# Create Logistic Regression model
model = LogisticRegression()

# Train the model
model.fit(X, y)

# Print coefficients
print("\nModel Coefficients:")
print(model.coef_)

# Print intercept
print("\nModel Intercept:")
print(model.intercept_)
```

### Task 3: Plot the Sigmoid Function

```python
import numpy as np
import matplotlib.pyplot as plt

# Generate values from -10 to 10
x = np.linspace(-10, 10, 100)

# Calculate sigmoid values
y = 1 / (1 + np.exp(-x))

# Plot sigmoid function
plt.plot(x, y, label="Sigmoid Function")

# Add threshold line
plt.axhline(y=0.5, color="red", linestyle="--",
            label="Threshold = 0.5")

# Add labels and title
plt.xlabel("Input Values")
plt.ylabel("Probability")
plt.title("Sigmoid Function")

plt.legend()
plt.grid()

plt.show()
```

### Task 4: Predict Viral or Not Viral using a Threshold

```python
# Create classification function
def classify_post(probability, threshold=0.5):

    if probability >= threshold:
        return "Viral"
    else:
        return "Not Viral"


# Test probabilities with threshold 0.5
print("Threshold = 0.5")

print(0.2, ":", classify_post(0.2))
print(0.6, ":", classify_post(0.6))
print(0.8, ":", classify_post(0.8))


# Test probabilities with threshold 0.7
print("\nThreshold = 0.7")

print(0.2, ":", classify_post(0.2, 0.7))
print(0.6, ":", classify_post(0.6, 0.7))
print(0.8, ":", classify_post(0.8, 0.7))
```
