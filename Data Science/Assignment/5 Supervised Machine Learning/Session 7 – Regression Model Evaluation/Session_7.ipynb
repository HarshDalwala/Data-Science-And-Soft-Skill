### Task 1: Calculate MSE and RMSE for Zomato delivery times

```python
import numpy as np
from sklearn.metrics import mean_squared_error

# Actual delivery times in minutes
actual = np.array([30, 40, 25, 50, 35, 45, 20, 55, 38, 42])

# Predicted delivery times in minutes
predicted = np.array([32, 38, 28, 48, 37, 43, 22, 52, 40, 45])

# Calculate MSE
mse = mean_squared_error(actual, predicted)

# Calculate RMSE
rmse = np.sqrt(mse)

print("MSE:", mse)
print("RMSE:", rmse)
```

### Task 2: Calculate MAE and R² for Flipkart product prices

```python
from sklearn.metrics import mean_absolute_error, r2_score

# Actual product prices
actual_prices = [500, 1000, 1500, 2000, 2500, 3000, 3500, 4000]

# Predicted product prices
predicted_prices = [520, 950, 1480, 2050, 2450, 3100, 3400, 3900]

# Create a function to calculate MAE and R2
def evaluate_model(actual, predicted):
    mae = mean_absolute_error(actual, predicted)
    r2 = r2_score(actual, predicted)

    print("MAE:", mae)
    print("R2 Score:", r2)

# Call the function
evaluate_model(actual_prices, predicted_prices)
```

### Task 3: Create a residual plot for movie ratings

```python
import matplotlib.pyplot as plt

# Actual movie ratings (scale 1 to 5)
actual_ratings = [4.5, 3.8, 4.2, 2.5, 3.0,
                  4.8, 3.5, 2.8, 4.0, 3.2,
                  4.6, 2.0, 3.7, 4.3, 2.7,
                  3.9, 4.1, 3.3, 4.7, 2.9]

# Predicted movie ratings
predicted_ratings = [4.3, 3.9, 4.0, 2.8, 3.2,
                     4.6, 3.6, 3.0, 3.8, 3.4,
                     4.5, 2.2, 3.5, 4.1, 2.9,
                     4.0, 4.2, 3.1, 4.5, 3.0]

# Calculate residuals
residuals = [
    actual - predicted
    for actual, predicted in zip(actual_ratings, predicted_ratings)
]

# Create residual plot
plt.scatter(predicted_ratings, residuals)

# Add zero reference line
plt.axhline(y=0, color='red', linestyle='--')

plt.xlabel("Predicted Movie Ratings")
plt.ylabel("Residuals")
plt.title("Residual Plot for Movie Ratings")

plt.show()
```

### Task 5: Compare two Spotify popularity prediction models

```python
import numpy as np
from sklearn.metrics import mean_squared_error
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import r2_score

# Actual Spotify popularity values
y_train = np.array([50, 60, 70, 80, 90, 40, 55, 65])
y_test = np.array([45, 58, 72, 83, 88, 48, 62, 75])

# Model A predictions
train_a = np.array([50, 60, 70, 80, 90, 40, 55, 65])
test_a = np.array([60, 70, 60, 95, 75, 65, 45, 90])

# Model B predictions
train_b = np.array([52, 58, 68, 82, 87, 43, 57, 63])
test_b = np.array([47, 60, 70, 80, 90, 46, 60, 73])


# Function to calculate evaluation metrics
def evaluate_model(actual, predicted):

    mse = mean_squared_error(actual, predicted)
    rmse = np.sqrt(mse)
    mae = mean_absolute_error(actual, predicted)
    r2 = r2_score(actual, predicted)

    print("MSE:", round(mse, 2))
    print("RMSE:", round(rmse, 2))
    print("MAE:", round(mae, 2))
    print("R2 Score:", round(r2, 2))


# Evaluate Model A
print("MODEL A - TRAINING DATA")
evaluate_model(y_train, train_a)

print("\nMODEL A - TEST DATA")
evaluate_model(y_test, test_a)


# Evaluate Model B
print("\nMODEL B - TRAINING DATA")
evaluate_model(y_train, train_b)

print("\nMODEL B - TEST DATA")
evaluate_model(y_test, test_b)
```
