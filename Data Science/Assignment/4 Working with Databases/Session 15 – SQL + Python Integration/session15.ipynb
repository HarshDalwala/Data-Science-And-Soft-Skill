```python
# ============================================
# SESSION 15: SQL + PYTHON INTEGRATION
# ============================================

# Import required libraries
import sqlite3
import pandas as pd


# ============================================
# TASK 1
# Create foodie.db and Restaurants table
# ============================================

# Connect to SQLite database
# If foodie.db does not exist, Python will create it
conn = sqlite3.connect("foodie.db")

# Create cursor
cursor = conn.cursor()

# Create Restaurants table
cursor.execute("""
CREATE TABLE IF NOT EXISTS Restaurants (
    id INTEGER PRIMARY KEY,
    name TEXT,
    cuisine TEXT,
    rating REAL
)
""")

# Save changes
conn.commit()

print("foodie.db database and Restaurants table created successfully.")


# ============================================
# TASK 2
# Insert 3 restaurants
# Fetch restaurants with rating above 4.0
# ============================================

# Insert sample restaurants
restaurants = [
    (1, "Swagat Restaurant", "South Indian", 4.5),
    (2, "Pizza Palace", "Italian", 4.2),
    (3, "Royal Chinese", "Chinese", 3.8)
]

cursor.executemany("""
INSERT OR REPLACE INTO Restaurants
(id, name, cuisine, rating)
VALUES (?, ?, ?, ?)
""", restaurants)

conn.commit()

print("\nRestaurants with rating above 4.0:")

# Fetch restaurants with rating > 4.0
cursor.execute("""
SELECT name
FROM Restaurants
WHERE rating > 4.0
""")

# Fetch all matching rows
results = cursor.fetchall()

# Print restaurant names
for row in results:
    print(row[0])


# ============================================
# TASK 3
# Load Restaurants table into Pandas DataFrame
# ============================================

query = "SELECT * FROM Restaurants"

df = pd.read_sql_query(query, conn)

print("\nComplete DataFrame:")
print(df)

print("\nTop 2 rows:")
print(df.head(2))


# ============================================
# TASK 4
# Add delivery_charge and final_rating
# ============================================

# Add delivery charge of 50 for all restaurants
df["delivery_charge"] = 50

# Add 0.1 to rating if cuisine is Italian
df["final_rating"] = df.apply(
    lambda row: row["rating"] + 0.1
    if row["cuisine"] == "Italian"
    else row["rating"],
    axis=1
)

print("\nDataFrame after adding new columns:")
print(df)


# ============================================
# TASK 5
# Find restaurants with rating above 4.5
# Save result as CSV
# ============================================

# Fetch restaurants with rating above 4.5
top_rated_query = """
SELECT *
FROM Restaurants
WHERE rating > 4.5
"""

# Load result into DataFrame
top_rated_df = pd.read_sql_query(top_rated_query, conn)

# Save as CSV
top_rated_df.to_csv(
    "top_rated_restaurants.csv",
    index=False
)

print("\nTop-rated restaurants:")
print(top_rated_df)

print("\nCSV file created: top_rated_restaurants.csv")


# ============================================
# Close database connection
# ============================================

conn.close()

print("\nDatabase connection closed.")
```
