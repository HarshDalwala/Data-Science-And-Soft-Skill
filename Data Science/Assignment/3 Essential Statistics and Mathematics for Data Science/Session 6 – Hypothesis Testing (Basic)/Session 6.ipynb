### Scenario
We want to test whether people spend more time on Instagram than YouTube daily.

### Null Hypothesis (H₀)
There is no significant difference, or people spend equal or less time on Instagram compared to YouTube.

**H₀: Instagram time ≤ YouTube time**

### Alternative Hypothesis (H₁)
People spend more time on Instagram than YouTube daily.

**H₁: Instagram time > YouTube time**

### Conclusion
Since our belief is that Instagram usage is higher than YouTube usage, this is a **one-tailed test**.

```python
from scipy.stats import ttest_ind_from_stats

# Group A - New homepage
mean_a = 350
std_a = 50
n_a = 30

# Group B - Old homepage
mean_b = 320
std_b = 50
n_b = 30

# Perform two-sample t-test
t_stat, p_value = ttest_ind_from_stats(
    mean_a, std_a, n_a,
    mean_b, std_b, n_b,
    equal_var=True
)

print("T-statistic:", round(t_stat, 2))
print("P-value:", round(p_value, 4))

# Decision
if p_value < 0.05:
    print("Reject the null hypothesis")
    print("The difference is statistically significant.")
else:
    print("Fail to reject the null hypothesis")
    print("The difference is not statistically significant.")
```

### Feature: Paytm Payment Button

I would test a redesigned payment button to check whether it improves the number of successful payments.

### Control Group (A)
Users see the **old payment button**.

### Variant Group (B)
Users see the **new redesigned payment button**.

### Metric
I would measure the **payment success rate**.

Payment Success Rate = (Successful Payments / Total Payment Attempts) × 100

### Null Hypothesis (H₀)
The new payment button does not improve the payment success rate.

**H₀: New button success rate ≤ Old button success rate**

### Alternative Hypothesis (H₁)
The new payment button improves the payment success rate.

**H₁: New button success rate > Old button success rate**

### Decision
If the p-value is less than 0.05, I would reject the null hypothesis and conclude that the new payment button has a statistically significant improvement.

In the A/B test, 520 out of 1,000 users clicked the new button, while 480 users clicked the old button.

The new button has a higher click rate (52%) compared to the old button (48%). However, this difference alone does not prove that the new button is better. We need to perform a statistical test and check the p-value.

If the p-value is less than 0.05, we reject the null hypothesis and consider the difference statistically significant. If the p-value is greater than or equal to 0.05, we fail to reject the null hypothesis and the difference may be due to chance.

Therefore, the p-value helps us decide whether the observed difference is statistically significant or likely due to random variation.

```python
from statsmodels.stats.proportion import proportions_ztest

# Clicks
clicks = [520, 480]

# Total users in each group
users = [1000, 1000]

# Two-proportion z-test
z_stat, p_value = proportions_ztest(clicks, users)

print("Z-statistic:", round(z_stat, 2))
print("P-value:", round(p_value, 4))

if p_value < 0.05:
    print("The difference is statistically significant.")
else:
    print("The difference may be due to chance.")
```
