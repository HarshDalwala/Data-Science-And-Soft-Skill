```python
import numpy as np
import matplotlib.pyplot as plt

# Generate 100 random numbers
data = np.random.normal(50, 10, 100)

# Plot histogram
plt.hist(data, bins=10, edgecolor="black")

plt.title("Normal Distribution")
plt.xlabel("Values")
plt.ylabel("Frequency")

plt.show()
```

I generated 100 random numbers using a normal distribution with a mean of 50 and a standard deviation of 10.

The histogram shows the approximate bell-shaped pattern of a normal distribution. Most values are close to the mean of 50, while fewer values occur farther away from the mean.

```python
import numpy as np
from scipy import stats

steps = [
    5000, 5200, 5500, 5600, 5800,
    5900, 6000, 6000, 6100, 6200,
    6300, 6400, 6500, 6600, 6700,
    6800, 6900, 7000, 7200, 7500
]

mean = np.mean(steps)
median = np.median(steps)
mode = stats.mode(steps, keepdims=True).mode[0]

print("Mean:", mean)
print("Median:", median)
print("Mode:", mode)
```

The mean is 6295, the median is 6250, and the mode is 6000.

These values are relatively close to each other, which suggests that the data is approximately symmetric and has some characteristics of a normal distribution.

In a perfectly normal distribution, the mean, median, and mode are approximately equal.

```python
import numpy as np

# Generate scores for 200 players
scores = np.random.normal(70, 15, 200)

# Count scores between 55 and 85
count = np.sum((scores >= 55) & (scores <= 85))

print("Number of players scoring between 55 and 85:", count)
print("Percentage:", round((count / 200) * 100, 2), "%")
```


I simulated the scores of 200 players using a normal distribution with a mean of 70 and a standard deviation of 15.

The range from 55 to 85 represents one standard deviation below and above the mean.

According to the 68-95-99.7 rule, approximately 68% of the players should score within this range. Because the data is randomly generated, the actual percentage may be slightly different from 68%.

### Example: Zomato Delivery Times

Suppose we collect the delivery times of many Zomato orders in a particular city.

For example, most orders may be delivered around 30 minutes, while only a few orders take much less or much more time.

If the delivery times form a roughly bell-shaped pattern around the average delivery time, a normal distribution can be a useful model.

Normal distribution can help Zomato understand the usual delivery time and identify unusually fast or delayed orders. It can also help estimate the percentage of orders that are delivered within a particular time range.
