### Explanation

A **population** is the complete group that we want to study.

A **sample** is a smaller group selected from the population to represent the larger group. Using a sample saves time and resources while still helping us understand the population.

```python
import random

# Create 1000 user IDs
users = [f"user{i}" for i in range(1, 1001)]

# Select 50 random users
sample_users = random.sample(users, 50)

# Print selected users
print("Selected 50 users:")
print(sample_users)
```
