## Task 1: Daily Actions and Data Types

| App | Daily Action | Data Generated | Data Type |
|---|---|---|---|
| Instagram | Like a post | Number of likes | Numerical |
| Instagram | Watch a Reel | Reel category | Categorical |
| Zomato | Give a restaurant rating | Rating (4.5) | Numerical |
| Zomato | Place an order | Food category | Categorical |
| Flipkart | Buy a product | Product price (₹999) | Numerical |

### Explanation
Numerical data represents numbers that can be counted or measured, such as likes, ratings, and prices. Categorical data represents labels or groups, such as Reel type or food category.

## Task 4: Statistical Terms for Spotify Playlist Analysis

If I analyze my Spotify playlist, I can use the following three statistical terms:

### 1. Mean
Mean is the average value of the data.

Example:
If I listen to songs for 30, 40, and 50 minutes:

Mean = (30 + 40 + 50) / 3 = 40 minutes

### 2. Median
Median is the middle value when the data is arranged in ascending or descending order.

Example:
Listening time = 20, 30, 50 minutes

Median = 30 minutes

### 3. Mode
Mode is the value that occurs most frequently in the data.

Example:
Favorite genres = Pop, Rock, Pop, Jazz, Pop

Mode = Pop

### Conclusion
Mean, median, and mode help us understand and summarize Spotify listening data easily.
