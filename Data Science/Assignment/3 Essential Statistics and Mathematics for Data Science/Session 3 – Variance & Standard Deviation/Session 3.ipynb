```python
steps = [4500, 7000, 5000, 8000, 4000, 9000, 6000]

# Step 1: Mean
mean = sum(steps) / len(steps)
print("Mean =", mean)

# Step 2: Show deviation and squared deviation
print("\nValue\tDeviation\tSquared Deviation")

squared_diff = []

for x in steps:
    diff = x - mean
    square = diff ** 2
    squared_diff.append(square)
    print(f"{x}\t{diff:.0f}\t\t{square:.0f}")

# Step 3: Variance
variance = sum(squared_diff) / len(steps)

print("\nSum of Squared Deviations =", sum(squared_diff))
print("Variance =", variance)
```

```python
import math

def calculate_standard_deviation(scores):
    mean = sum(scores) / len(scores)

    squared_differences = []

    for score in scores:
        squared_differences.append((score - mean) ** 2)

    variance = sum(squared_differences) / len(scores)

    standard_deviation = math.sqrt(variance)

    return round(standard_deviation, 2)


scores = [75, 80, 85, 90, 70]

print("Standard Deviation:", calculate_standard_deviation(scores))
```

```python
import math

friend_a = [200, 200, 200, 200]
friend_b = [100, 300, 150, 400, 50]

def standard_deviation(data):
    mean = sum(data) / len(data)
    
    variance = sum((x - mean) ** 2 for x in data) / len(data)
    
    return math.sqrt(variance)

sd_a = standard_deviation(friend_a)
sd_b = standard_deviation(friend_b)

print("Friend A Standard Deviation:", round(sd_a, 2))
print("Friend B Standard Deviation:", round(sd_b, 2))
```

```python
import math

salaries = [28000, 29000, 27000, 60000, 26500, 27500]

# Calculate mean
mean = sum(salaries) / len(salaries)

# Calculate variance
variance = sum((salary - mean) ** 2 for salary in salaries) / len(salaries)

# Calculate standard deviation
standard_deviation = math.sqrt(variance)

print("Mean Salary:", round(mean, 2))
print("Variance:", round(variance, 2))
print("Standard Deviation:", round(standard_deviation, 2))
```
