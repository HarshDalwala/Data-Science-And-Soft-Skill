```python
import matplotlib.pyplot as plt

orders = [12, 15, 17, 14, 13, 16, 200, 18, 14, 15]

plt.plot(orders, marker='o')
plt.title("Swiggy Daily Order Counts")
plt.xlabel("Day")
plt.ylabel("Number of Orders")
plt.show()
```

```python
followers = [5, 7, 8, 8, 9, 10, 12, 15, 95]

mean = sum(followers) / len(followers)

followers.sort()
median = followers[len(followers) // 2]

print("Mean:", mean, "thousand")
print("Median:", median, "thousand")
```

```python
views = [500, 800, 1000, 1200, 1500, 2000, 2500, 3000, 50000]

mean = sum(views) / len(views)

views.sort()
median = views[len(views) // 2]

print("Mean views:", mean)
print("Median views:", median)
```

```python
import matplotlib.pyplot as plt

ratings = [3, 3, 4, 4, 4, 5, 5, 5, 5, 5]

plt.hist(ratings, bins=[2.5, 3.5, 4.5, 5.5], edgecolor="black")
plt.title("Flipkart Product Ratings")
plt.xlabel("Rating")
plt.ylabel("Frequency")
plt.xticks([3, 4, 5])
plt.show()
```
